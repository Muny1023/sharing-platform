# AI 服务控制

请先在 VS Code 右上角选择 `langchain1.2` 内核，然后运行下面的启动单元。

```python
import os
import sys
import subprocess
import socket
from pathlib import Path

ai_dirs = [Path.cwd(), Path.cwd() / 'my-project-ai']
ai_dir = next(path for path in ai_dirs if (path / 'app').is_dir())
sys.path.insert(0, str(ai_dir))

def ai_port_in_use():
    with socket.socket() as sock:
        return sock.connect_ex(('127.0.0.1', 8000)) == 0

if ai_port_in_use():
    ai_process = None
    print('AI 服务已经在运行：http://127.0.0.1:8000')
elif 'ai_process' not in globals() or ai_process is None or ai_process.poll() is not None:
    ai_process = subprocess.Popen(
        [sys.executable, str(ai_dir / 'app' / 'main.py')],
        cwd=str(ai_dir),
        env=os.environ.copy(),
    )
    print(f'AI 服务已启动：http://127.0.0.1:8000 (PID: {ai_process.pid})')
else:
    print(f'AI 服务已经在运行 (PID: {ai_process.pid})')
```

```python
if 'ai_process' in globals() and ai_process is not None and ai_process.poll() is None:
    ai_process.terminate()
    try:
        ai_process.wait(timeout=5)
    except subprocess.TimeoutExpired:
        ai_process.kill()
    print('AI 服务已停止')
else:
    print('当前没有启动的 AI 服务')
```
